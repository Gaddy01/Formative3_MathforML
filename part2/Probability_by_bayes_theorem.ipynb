```python
total_reviews = len(data)
positive_reviews = len(data[data['sentiment'] == "positive"])
P_positive = positive_reviews / total_reviews
print(f"P(positive) = {P_positive}")
```

```python
Positive_Keywords = ["excellent", "wonderful", "amazing", "love"]
negative_keywords = ["terrible", "awful", "boring", "worst"]
```

### Keyword Selection Justification

The selected keywords represent strong emotional expressions commonly
associated with sentiment polarity in movie reviews.

Positive keywords such as "excellent", "wonderful", "amazing",
and "love" typically reflect reviewer satisfaction and approval.
These words are subjective evaluative terms rather than neutral descriptions,
making them strong indicators of positive sentiment.

Negative keywords such as "terrible", "awful", "boring",
and "worst" represent dissatisfaction or criticism and are
frequently used in negative reviews.

```python
data['review'] = data['review'].str.lower()
```

```python
# we calculate :

"""P(Positive | Keyword)"""

# Bayes Theorem

"""P(Positive | Keyword) = P(Keyword |Positive) * P(Positive) / P(keyword)"""
```

```python
# Bayes Implementation by Code
```

```python
# Keyword count

def keyword_count(keyword, data):
    return len(data[data['review'].str.contains(keyword)])
```

```python
# Likelihood

def likelihood(keyword):
    positive_reviews = data[data['sentiment'] == 'positive']
    return keyword_count(keyword, positive_reviews) / len(positive_reviews)
```

```python
# Marginal Probability

def marginal(keyword):
    return keyword_count(keyword, data) / len(data)
```

```python
# Bayes Theorem

def posterior(keyword):
    P_k_given_positive = likelihood(keyword)
    P_k = marginal(keyword)

    if P_k == 0:
        return 0

    return P_k_given_positive * P_positive / P_k
```

### Bayes' Theorem

We compute the probability that a review is positive given
the presence of a keyword using Bayes' Theorem:

P(Positive | keyword) =
(P(keyword | Positive) × P(Positive)) / P(keyword)

Where:
- P(Positive) is the prior probability
- P(keyword | Positive) is the likelihood
- P(keyword) is the marginal probability
- P(Positive | keyword) is the posterior probability

```python
# computing all Probabilities

results = []

for word in Positive_Keywords:
    results.append([
        word,
        P_positive,
        likelihood(word),
        marginal(word),
        posterior(word)
    ])
```

```python
# Results table display

result_data = pd.DataFrame(
    results,
    columns=[
        "Keyword",
        "Prior P(Positive)",
        "Likelihood P(keyword|Positive)",
        "Marginal P(keyword)",
        "Posterior P(Positive|keyword)"
    ]
)

result_data = result_data.round(4)
result_data
```

### Interpretation of Results

Keywords such as "wonderful" and "excellent"
produce posterior probabilities above 0.80,
indicating a strong association with positive sentiment.

This demonstrates that the presence of emotionally
positive words significantly increases the probability
that a review belongs to the positive class.
